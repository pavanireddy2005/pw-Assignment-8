Question 4: Explain how props support React’s one-way data flow.

Answer:
Props (properties) are used to pass data from a parent component to a child component in React.

React follows a one-way data flow, meaning:

Data moves only from parent to child.

Child components cannot directly modify parent data.

This makes applications predictable and easier to debug.

```python
Example:
function Child(props) {
  return <h1>{props.message}</h1>;
}

function Parent() {
  return <Child message="Hello from Parent" />;
}
```

Question 5: Create a functional component ClickMeButton that renders a button. It accepts onClick and label props. When clicked, it executes the passed onClick handler.

```python
function ClickMeButton(props) {
  return (
    <button onClick={props.onClick}>
      {props.label}
    </button>
  );
}

export default ClickMeButton;
```

Question 6: Given const fruits = ['Apple', 'Banana', 'Cherry'], write a component FruitList that renders an unordered list of these fruits using .map() and includes a key.

```python
function FruitList() {
  const fruits = ['Apple', 'Banana', 'Cherry'];

  return (
    <ul>
      {fruits.map((fruit, index) => (
        <li key={index}>{fruit}</li>
      ))}
    </ul>
  );
}

export default FruitList;
```

Question 7: Why is the key attribute important when rendering lists in React? What goes wrong if keys are missing or misuse array index?

Answer:
The key attribute helps React identify which list items have changed, been added, or removed.

Importance of keys:
Improves rendering performance.

Helps React efficiently update the DOM.

Prevents unnecessary re-rendering.
Problems if keys are missing:
React shows warnings.

UI updates may become inefficient.

Incorrect components may get updated.
Problems when using array index as key:

Using array index as a key can cause issues when:

Items are added or removed.

List order changes.

This may lead to:

Incorrect UI rendering.

State mismatch between components.

Therefore, unique and stable keys such as database IDs should be used whenever possible.

Question 8: Write a component UserList that receives a users prop—an array of objects each with id and name. Render each user’s name in a <p> tag, using map() and key.

```python
function UserList(props) {
  return (
    <div>
      {props.users.map((user) => (
        <p key={user.id}>
          {user.name}
        </p>
      ))}
    </div>
  );
}

export default UserList;
```

Question 9: Create an App component that holds an array of numbers, passes them to a NumberList child component for rendering, and passes a click handler function to log the clicked number. Show one-way data flow and event handling.

```python
function NumberList(props) {
  return (
    <div>
      {props.numbers.map((num) => (
        <button
          key={num}
          onClick={() => props.onNumberClick(num)}
        >
          {num}
        </button>
      ))}
    </div>
  );
}

function App() {
  const numbers = [1, 2, 3, 4, 5];

  function handleNumberClick(number) {
    console.log("Clicked Number:", number);
  }

  return (
    <div>
      <h1>Number List</h1>

      <NumberList
        numbers={numbers}
        onNumberClick={handleNumberClick}
      />
    </div>
  );
}

export default App;
```

Explanation:
App is the parent component.

numbers array is passed to NumberList using props.

Click handler is also passed as a prop.

Clicking a button triggers the function in the parent component.

This demonstrates one-way data flow and event handling in React.

Question 10: Design a component ProductList that receives products (array of objects with id, name, price). Render a list of product names and prices using .map(). Each item when clicked should call a prop onProductSelect(product). Include the proper key.

```python
function ProductList(props) {
  return (
    <div>
      {props.products.map((product) => (
        <div
          key={product.id}
          onClick={() => props.onProductSelect(product)}
          style={{
            cursor: "pointer",
            marginBottom: "10px"
          }}
        >
          <h3>{product.name}</h3>
          <p>Price: ${product.price}</p>
        </div>
      ))}
    </div>
  );
}

export default ProductList;
```
